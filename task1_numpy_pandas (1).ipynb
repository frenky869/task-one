## 2. NumPy Fundamentals

### 2.1 Array Creation

```python
# Create arrays from DataFrame columns
ages = df['Age'].dropna().to_numpy()
fares = df['Fare'].to_numpy()

# Various creation methods
zeros_arr = np.zeros(5)
ones_arr = np.ones((2, 3))
range_arr = np.arange(0, 20, 2)
linspace_arr = np.linspace(0, 1, 5)

print("Ages array (first 10):", ages[:10])
print("Zeros:", zeros_arr)
print("Ones (2x3):\n", ones_arr)
print("Arange:", range_arr)
print("Linspace:", linspace_arr)
```

### 2.2 Indexing & Slicing

```python
# Basic indexing
print("First age:", ages[0])
print("Last age:", ages[-1])

# Slicing
print("First 5 ages:", ages[:5])
print("Every 3rd age (first 15):", ages[:15:3])
print("Reversed first 5:", ages[4::-1])

# Boolean (fancy) indexing
adult_ages = ages[ages >= 18]
print(f"\nTotal passengers with known age: {len(ages)}")
print(f"Adults (18+): {len(adult_ages)}")
print(f"Minors (<18): {len(ages) - len(adult_ages)}")

# 2D indexing example
matrix = np.arange(12).reshape(3, 4)
print("\nMatrix:\n", matrix)
print("Row 1:", matrix[1])
print("Column 2:", matrix[:, 2])
print("Sub-block:\n", matrix[0:2, 1:3])
```

### 2.3 Broadcasting

```python
# Broadcasting: scalar operations applied elementwise without explicit loops
fare_in_cents = fares * 100
fares_normalized = (fares - fares.mean()) / fares.std()

print("Original fares (first 5):", fares[:5])
print("Fares in cents (first 5):", fare_in_cents[:5])
print("Normalized fares (first 5):", np.round(fares_normalized[:5], 3))

# Broadcasting a (3,4) matrix with a (4,) vector
matrix = np.arange(12).reshape(3, 4)
vector = np.array([1, 2, 3, 4])
print("\nMatrix + vector (broadcasting):\n", matrix + vector)

# Vectorized vs loop comparison (why vectorization matters)
import time

big_arr = np.random.rand(1_000_000)

start = time.time()
loop_result = [x * 2 for x in big_arr]
loop_time = time.time() - start

start = time.time()
vec_result = big_arr * 2
vec_time = time.time() - start

print(f"\nPython loop time: {loop_time:.4f}s")
print(f"Vectorized (broadcasting) time: {vec_time:.4f}s")
print(f"Speedup: {loop_time / vec_time:.1f}x")
```

## 3. Pandas DataFrame Operations

### 3.1 Filtering with `loc[]` / `iloc[]`

```python
# label-based selection with loc
women_first_class = df.loc[(df['Sex'] == 'female') & (df['Pclass'] == 1),
                             ['Sex', 'Pclass', 'Age', 'Fare', 'Survived']]
print("Women in 1st class (via .loc):")
women_first_class.head()
```

```python
# position-based selection with iloc
print("First 5 rows, columns 0-4 (via .iloc):")
df.iloc[:5, 0:5]
```

```python
# Filtering: survivors older than 60
elderly_survivors = df.loc[(df['age'] > 60) & (df['survived'] == 1)]
print(f"Survivors older than 60: {len(elderly_survivors)}")
elderly_survivors[['sex', 'age', 'pclass', 'fare']]
```

### 3.2 Grouping & Aggregation (`groupby()` + `agg()`)

```python
# Aggregation 1: Survival rate by passenger class
survival_by_class = df.groupby('Pclass')['Survived'].agg(['mean', 'count'])
survival_by_class.columns = ['Survival_rate', 'Passenger_count']
print("Survival rate by class:")
survival_by_class
```

```python
# Aggregation 2: Multiple statistics on fare and age, grouped by sex
sex_stats = df.groupby('Sex').agg(
    avg_fare=('Fare', 'mean'),
    max_fare=('Fare', 'max'),
    avg_age=('Age', 'mean'),
    survival_rate=('Survived', 'mean')
)
print("Stats grouped by sex:")
sex_stats
```

```python
# Aggregation 3: Grouped by class AND sex (multi-index groupby)
class_sex_survival = df.groupby(['Pclass', 'Sex'])['Survived'].agg(['mean', 'sum', 'count'])
class_sex_survival.columns = ['Survival_rate', 'Survivors', 'Total']
print("Survival by class and sex:")
class_sex_survival
```

### 3.3 Merging / Joining

```python

# Lookup table
class_fare_lookup = pd.DataFrame({
    'Pclass': [1, 2, 3],
    'Class_label': ['First Class', 'Second Class', 'Third Class'],
    'Deck_priority': ['A-E', 'D-F', 'E-G']
})

# Merge on 'Pclass' (note the capital P)
merged = pd.merge(df, class_fare_lookup, on='Pclass', how='left')

print("Merged DataFrame (added Class_label, Deck_priority):")
print(merged[['Pclass', 'Class_label', 'Deck_priority', 'Survived']].head())
```

```python
# Build partial lookup excluding class 3
partial_lookup = class_fare_lookup[class_fare_lookup['Pclass'] != 3]

# Inner vs Left join
inner_merged = pd.merge(df, partial_lookup, on='Pclass', how='inner')
left_merged = pd.merge(df, partial_lookup, on='Pclass', how='left')

print(f"Original rows: {len(df)}")
print(f"Inner join rows (drops Pclass=3): {len(inner_merged)}")
print(f"Left join rows (keeps all, NaN for missing): {len(left_merged)}")

print(f"\nNaNs introduced by left join in 'Class_label': {left_merged['Class_label'].isna().sum()}")
```

## 4. Summary of Key Findings

```python
# Survival rate by class
survival_by_class = (
    df.groupby("Pclass")["Survived"]
      .mean()
      .reset_index(name="Survival_rate")
      .set_index("Pclass")
)

# Survival rate by sex
sex_stats = (
    df.groupby("Sex")["Survived"]
      .mean()
      .reset_index(name="Survival_rate")
      .set_index("Sex")
)
```

```python
print("=== KEY FINDINGS ===")
print(f"1. Overall survival rate: {df['Survived'].mean():.1%}")
print(f"2. 1st class survival rate: {survival_by_class.loc[1, 'Survival_rate']:.1%} "
      f"vs 3rd class: {survival_by_class.loc[3, 'Survival_rate']:.1%}")
print(f"3. Female survival rate: {sex_stats.loc['female', 'Survival_rate']:.1%} "
      f"vs male: {sex_stats.loc['male', 'Survival_rate']:.1%}")
print(f"4. Average fare: ${df['Fare'].mean():.2f}, Median: ${df['Fare'].median():.2f}")
print(f"5. Missing age data: {df['Age'].isna().sum()} out of {len(df)} passengers "
      f"({df['Age'].isna().mean():.1%})")
```
